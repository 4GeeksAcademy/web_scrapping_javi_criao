```python
import os
from bs4 import BeautifulSoup
import requests
import time
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import random
import re
```

```python
url = "https://books.toscrape.com/catalogue/category/books_1/index.html"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
book_elements = soup.find_all('article', class_='product_pod')
un_libro_azar = random.choice(book_elements)
print(f"Libro aleatorio capturado: {un_libro_azar.h3.a['title']}")
```

```python
datos_crudos = [] 

for book in book_elements:
    titulo = book.h3.a['title']
    precio_texto = book.find('p', class_='price_color').get_text()
    
    precio_solo_numeros = "".join(re.findall(r'[0-9.]', precio_texto))
    
    if precio_solo_numeros != "":
        datos_crudos.append({'Titulo': titulo, 'Precio': precio_solo_numeros})

df = pd.DataFrame(datos_crudos)

df['Precio'] = pd.to_numeric(df['Precio'])

print("Limpieza total completada")
print(df.head())
```

```python
conexion = sqlite3.connect("biblioteca_libros.db")

df.to_sql('libros', conexion, if_exists='replace', index=False)

conexion.commit()
conexion.close()

print("¡Paso 5 completado! Se ha creado el archivo 'biblioteca_libros.db' con tus datos.")
```
